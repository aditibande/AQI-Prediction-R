```python
plt.figure(figsize=(4,4))
colors=['#FF6347','#4682B4']
sns.set_style('whitegrid')
sns.countplot(x='species',data=df,palette="Set1")
```

```python
plt.figure(figsize=(6,4))
sns.histplot(data=df,x='petal_length',bins=20,kde=True,color='skyblue')
plt.title('Distribution of Petal length')
plt.xlabel('Petal length (cm)')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(7,4.5))
#create the histogram color-coded by species
sns.histplot(
    data=df,
    x='petal_length',
    hue='species',
    bins=20,
    kde=True,
    multiple ='stack',
    palette='muted'
    )
#add descriptive plot titles and axes labels
plt.title('petal ength distribution by species')
plt.xlabel('petal length (cm)')
plt.ylabel('count')

#cleanly render and display the plot window
plt.show()
```
